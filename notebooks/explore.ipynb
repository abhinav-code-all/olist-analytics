```python
import pandas as pd
from pathlib import Path

raw = Path(r"C:\Projects\olist-analytics\data\raw")
dfs = {f.stem: pd.read_csv(f) for f in raw.glob("*.csv")}

for name, df in dfs.items():
    print(f"\n=== {name} ===")
    print("rows, cols:", df.shape)
    print("nulls:\n", df.isna().sum()[df.isna().sum() > 0])
```

```python
for name in ["olist_order_payments_dataset", "olist_order_reviews_dataset",
             "olist_products_dataset", "olist_sellers_dataset"]:
    df = dfs[name]
    n = df.isna().sum()
    print(f"\n=== {name} === rows, cols: {df.shape}")
    print(n[n > 0].to_string() if n.any() else "no nulls")

print("\n=== order status ===")
print(dfs["olist_orders_dataset"]["order_status"].value_counts().to_string())
```
